```python
from torchdiffeq import odeint
import torch

func = lambda t, y: -y    # dy/dt = -y
y0 = torch.tensor([1.0])
t = torch.linspace(0, 10, 100)

res = odeint(func, y0, t)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(t, res)
plt.plot(t, torch.exp(-t), 'r--')  # Plot the analytical solution for comparison  
```
